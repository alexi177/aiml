```python
# Install dependencies as needed:
# pip install kagglehub[pandas-datasets]
import kagglehub
from kagglehub import KaggleDatasetAdapter
import pandas as pd

# Set the path to the file you'd like to load
file_path = "Sports.csv"

# Load the latest version
df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "aarchikunchal/sports",
  file_path,
  # Provide any additional arguments like
  # sql_query or pandas_kwargs. See the
  # documenation for more information:
  # https://github.com/Kaggle/kagglehub/blob/main/README.md#kaggledatasetadapterpandas
)

print("First 5 records:", df.head())

print("==================================================================================================")
print("2)")



print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
print("\nData types and missing values:")
df.info()


print("==================================================================================================")
print("3)")

print("First 5 records:")
display(df.head())

print("\nLast 5 records:")
display(df.tail())

print("\nRandom 5 records:")
display(df.sample(5))



print("==================================================================================================")
print("4)")

print("Missing values per column:")
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])

if missing_values.sum() == 0:
    print("No missing values found in the dataset.")
else:
    print("\nFurther investigation needed to handle missing values.")


print("==================================================================================================")
print("5)")

num_duplicates = df.duplicated().sum()
print(f"Number of duplicate records: {num_duplicates}")

if num_duplicates > 0:
    df.drop_duplicates(inplace=True)
    print("Duplicate records removed.")
    print(f"Number of rows after removing duplicates: {df.shape[0]}")
else:
    print("No duplicate records found.")

print("==================================================================================================")
print("6)")

import matplotlib.pyplot as plt
import seaborn as sns

# Identify numerical columns for outlier detection
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

print("Numerical columns for outlier detection:", numerical_cols.tolist())

# Create box plots for numerical columns
plt.figure(figsize=(15, 5 * len(numerical_cols)))
for i, col in enumerate(numerical_cols):
    plt.subplot(len(numerical_cols), 2, 2*i + 1)
    sns.boxplot(y=df[col])
    plt.title(f'Box plot of {col}')
    plt.ylabel(col)

    # Display descriptive statistics for each numerical column
    plt.subplot(len(numerical_cols), 2, 2*i + 2)
    plt.text(0.1, 0.5, df[col].describe().to_string(), {'fontsize': 12}, fontproperties = 'monospace')
    plt.title(f'Descriptive Statistics for {col}')
    plt.axis('off')

plt.tight_layout()
plt.show()

print("==================================================================================================")
print("7)")

# Based on the box plots, we can decide how to handle outliers.
# Common strategies include:
# 1. Removal: If outliers are few and clearly erroneous.
# 2. Transformation: Using log transformation or other power transformations to reduce the skewness caused by outliers.
# 3. Imputation: Replacing outliers with a more representative value (e.g., mean, median, or a specific percentile).
# 4. Capping/Winsorization: Limiting extreme values to a certain percentile.

# For this dataset, let's consider capping/winsorization for 'Runs' and 'Wickets' if necessary,
# or simply noting them if they represent legitimate but extreme player performances.

# Example: Capping 'Runs' at the 99th percentile (if deemed an outlier)
# Q1 = df['Runs'].quantile(0.25)
# Q3 = df['Runs'].quantile(0.75)
# IQR = Q3 - Q1
# upper_bound = Q3 + 1.5 * IQR
# df['Runs'] = df['Runs'].clip(upper=upper_bound)

print("Outlier handling strategy will depend on the specific context and domain knowledge.")
print("For this dataset, 'Runs' and 'Wickets' show some higher values, but they seem to represent exceptional player performances rather than errors. We will proceed without explicit outlier removal/capping for these columns at this stage, assuming they are valid data points.")


print("==================================================================================================")
print("8)")

import pandas as pd
from sklearn.preprocessing import LabelEncoder


categorical_cols_to_encode = ['Team', 'Role']

print(f"Categorical columns to be encoded: {categorical_cols_to_encode}")

# ---------- Label Encoding ----------
print("\nApplying Label Encoding...")

df_label_encoded_demo = df.copy()

for col in categorical_cols_to_encode:
    le = LabelEncoder()
    df_label_encoded_demo[f'{col}_LabelEncoded'] = le.fit_transform(df_label_encoded_demo[col])

display(df_label_encoded_demo.head())

# ---------- One-Hot Encoding ----------
print("\nApplying One-Hot Encoding...")

df_one_hot_encoded_demo = pd.get_dummies(
    df.copy(),
    columns=categorical_cols_to_encode,
    drop_first=True,
    dtype=int
)

display(df_one_hot_encoded_demo.head())

df_processed = df_one_hot_encoded_demo.copy()

print("==================================================================================================")
print("9)")


from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Identify numerical columns for scaling
# These are the original numerical columns that have varying magnitudes and would benefit from scaling.
numerical_cols_for_scaling = ['Matches', 'Runs', 'Wickets']

print(f"Numerical columns to be scaled: {numerical_cols_for_scaling}")

# Make copies of the processed DataFrame (which is now df_processed from previous step)
df_minmax_scaled = df_processed.copy()
df_standard_scaled = df_processed.copy()

# --- Min-Max Normalization ---
print("\nApplying Min-Max Normalization...")
min_max_scaler = MinMaxScaler()
# Apply scaling only to the specified numerical columns
df_minmax_scaled[numerical_cols_for_scaling] = min_max_scaler.fit_transform(df_minmax_scaled[numerical_cols_for_scaling])
print("Min-Max Scaled DataFrame head:")
display(df_minmax_scaled.head())

# --- Standardization (Z-score Scaling) ---
print("\nApplying Standardization (Z-score Scaling)...")
standard_scaler = StandardScaler()
# Apply scaling only to the specified numerical columns
df_standard_scaled[numerical_cols_for_scaling] = standard_scaler.fit_transform(df_standard_scaled[numerical_cols_for_scaling])
print("Standard Scaled DataFrame head:")
display(df_standard_scaled.head())

# We will use df_minmax_scaled for subsequent correlation analysis as it also contains the one-hot encoded features.


print("==================================================================================================")
print("10,11)")


import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical and one-hot encoded categorical columns for correlation analysis
# We will use df_minmax_scaled for consistency, which includes scaled numerical columns and one-hot encoded columns.
correlation_features = ['Matches', 'Runs', 'Wickets', 'Team_India', 'Team_New Zealand', 'Team_South Africa', 'Role_Bowler', 'Role_Wicket-Keeper']

# Ensure all correlation_features exist in df_minmax_scaled
missing_corr_features = [col for col in correlation_features if col not in df_minmax_scaled.columns]
if missing_corr_features:
    print(f"Warning: The following correlation features are missing from the DataFrame: {missing_corr_features}")
    # Filter to only include existing features
    correlation_features = [col for col in correlation_features if col in df_minmax_scaled.columns]

print(f"Features selected for correlation analysis: {correlation_features}")

# Calculate the correlation matrix
correlation_matrix = df_minmax_scaled[correlation_features].corr()

print("\nCorrelation Matrix:")
display(correlation_matrix)

# Generate a heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Selected Features')
plt.show()


print("==================================================================================================")
print("12)")

# Create 'Runs_Per_Match' feature
df_processed['Runs_Per_Match'] = df_processed['Runs'] / df_processed['Matches']
# Handle potential division by zero (if 'Matches' is 0, 'Runs_Per_Match' should be 0)
df_processed['Runs_Per_Match'].fillna(0, inplace=True)

# Create 'Wickets_Per_Match' feature
df_processed['Wickets_Per_Match'] = df_processed['Wickets'] / df_processed['Matches']
# Handle potential division by zero (if 'Matches' is 0, 'Wickets_Per_Match' should be 0)
df_processed['Wickets_Per_Match'].fillna(0, inplace=True)

print("DataFrame with new engineered features (Runs_Per_Match, Wickets_Per_Match):")
display(df_processed.head())

# Also update the scaled DataFrames with these new features for consistency in future steps
df_minmax_scaled['Runs_Per_Match'] = df_minmax_scaled['Runs'] / df_minmax_scaled['Matches']
df_minmax_scaled['Runs_Per_Match'].fillna(0, inplace=True)
df_minmax_scaled['Wickets_Per_Match'] = df_minmax_scaled['Wickets'] / df_minmax_scaled['Matches']
df_minmax_scaled['Wickets_Per_Match'].fillna(0, inplace=True)

df_standard_scaled['Runs_Per_Match'] = df_standard_scaled['Runs'] / df_standard_scaled['Matches']
df_standard_scaled['Runs_Per_Match'].fillna(0, inplace=True)
df_standard_scaled['Wickets_Per_Match'] = df_standard_scaled['Wickets'] / df_standard_scaled['Matches']
df_standard_scaled['Wickets_Per_Match'].fillna(0, inplace=True)




print("==================================================================================================")
print("13)")

print("Current data types in df_processed:")
df_processed.info()

print("\nBased on the `df.info()` output, the numerical columns ('Matches', 'Runs', 'Wickets', 'Runs_Per_Match', 'Wickets_Per_Match') are already in appropriate numerical formats (int64 or float64). The one-hot encoded columns are integers (0 or 1). 'PlayerID' and 'Name' are 'object' (string), which is suitable as they are identifiers and not used for numerical computations. No further explicit type conversions (like string to datetime) are immediately necessary for the current set of features and the planned analysis.")


print("==================================================================================================")
print("14)")



print("Summary of inconsistent and noisy data handling:")
print("  - **Missing Values:** Checked in step 4; no missing values were found.")
print("  - **Duplicate Records:** Checked and removed in step 5; no duplicates were found.")
print("  - **Outliers:** Detected and visualized in step 6. For 'Runs' and 'Wickets', the identified high values were deemed legitimate player performances rather than errors, and no aggressive capping/removal was applied (step 7).")
print("  - **Categorical Inconsistencies:** Handled during Label and One-Hot Encoding in step 8, which standardizes categories.")
print("  - **Numerical Noise:** Feature scaling (Min-Max and Standardization) in step 9 can help mitigate the impact of variations in numerical ranges.")
print("\nAt this stage, based on the general data cleaning steps, no further obvious inconsistencies or significant noise requiring additional cleaning techniques have been identified. The data appears reasonably clean for the next steps.")



print("==================================================================================================")
print("15)")

from sklearn.model_selection import train_test_split

# ==========================================
# Define Target Variable
# ==========================================

y = df_processed["Runs"]

# ==========================================
# Define Feature Variables
# ==========================================

# Remove only columns that actually exist.
# 'Team' and 'Role' no longer exist because they were
# converted into one-hot encoded columns.

drop_cols = ["PlayerID", "Name", "Runs"]

# Keep only columns that exist
drop_cols = [col for col in drop_cols if col in df_processed.columns]

X = df_processed.drop(columns=drop_cols)

print("\nFeatures (X):")
display(X.head())

print("\nTarget (y):")
display(y.head())

# ==========================================
# Train-Test Split
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTrain-Test Split Successful")
print("--------------------------------")
print("X_train shape :", X_train.shape)
print("X_test shape  :", X_test.shape)
print("y_train shape :", y_train.shape)
print("y_test shape  :", y_test.shape)


print("==================================================================================================")
print("16)")

print("Data balancing techniques like random oversampling or SMOTE are typically used for classification tasks to handle imbalanced classes. Since our current objective (predicting 'Runs') is a regression task, these techniques are not directly applicable.")
print("\nIf a classification task were intended (e.g., classifying players into 'high runs scorer' vs. 'low runs scorer'), then data balancing would be a relevant step after defining such a categorical target variable.")


print("==================================================================================================")
print("17)")

import matplotlib.pyplot as plt
import seaborn as sns

numerical_cols_to_visualize = ['Matches', 'Runs', 'Wickets']

plt.figure(figsize=(18, 12))

for i, col in enumerate(numerical_cols_to_visualize):
    # Original distribution
    plt.subplot(len(numerical_cols_to_visualize), 3, i * 3 + 1)
    sns.histplot(df[col], kde=True, bins=5)
    plt.title(f'Original {col} Distribution')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    # Min-Max Scaled distribution
    plt.subplot(len(numerical_cols_to_visualize), 3, i * 3 + 2)
    sns.histplot(df_minmax_scaled[col], kde=True, bins=5)
    plt.title(f'Min-Max Scaled {col} Distribution')
    plt.xlabel(f'{col} (Min-Max Scaled)')
    plt.ylabel('Frequency')

    # Standard Scaled distribution
    plt.subplot(len(numerical_cols_to_visualize), 3, i * 3 + 3)
    sns.histplot(df_standard_scaled[col], kde=True, bins=5)
    plt.title(f'Standard Scaled {col} Distribution')
    plt.xlabel(f'{col} (Standard Scaled)')
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()



print("==================================================================================================")
print("18)")




print("==================================================================================================")
print("19)")


output_file_path = 'cleaned_sports_data.csv'
df_processed.to_csv(output_file_path, index=False)
print(f"Cleaned and preprocessed dataset saved to '{output_file_path}'")

print("==================================================================================================")
print("20)")
```

### Summary of Preprocessing Changes:

1.  **Initial Data Loading and Inspection (Steps 2 & 3):**
    *   The dataset `df` was loaded, containing 10 rows and 7 columns. Initial data types were identified, and basic descriptive statistics were reviewed. No missing values were found at this stage.

2.  **Missing Values (Step 4):**
    *   A thorough check for missing values across all columns confirmed that there were no `null` entries in the dataset. Thus, no imputation or deletion of missing values was required.

3.  **Duplicate Records (Step 5):**
    *   The dataset was checked for duplicate rows. No duplicate records were found, meaning every row in the dataset represented a unique entry. Therefore, no rows were removed due to duplication.

4.  **Outlier Detection and Handling (Steps 6 & 7):**
    *   Outliers in numerical features ('Matches', 'Runs', 'Wickets') were detected and visualized using box plots. While some high values were observed, they were deemed legitimate player performances rather than errors (e.g., high runs or wickets by star players). Based on domain understanding, it was decided not to aggressively remove or cap these values, preserving the original distribution of these performance metrics.

5.  **Categorical Variable Encoding (Step 8):**
    *   The categorical columns ('Team', 'Role') were converted into numerical formats.
        *   **Label Encoding:** Demonstrated for 'Team' and 'Role' to show ordinal mapping (though not used in the final `df_processed`).
        *   **One-Hot Encoding:** Applied to 'Team' and 'Role', creating new binary columns (e.g., `Team_India`, `Role_Bowler`). The original categorical columns were dropped to avoid multicollinearity and prepare the data for models. The resulting DataFrame, `df_processed`, now contains these one-hot encoded features.

6.  **Feature Scaling (Step 9):**
    *   Numerical features ('Matches', 'Runs', 'Wickets') underwent feature scaling to normalize their ranges:
        *   **Min-Max Normalization:** Scaled values to a range between 0 and 1, creating `df_minmax_scaled`.
        *   **Standardization (Z-score Scaling):** Transformed values to have a mean of 0 and a standard deviation of 1, creating `df_standard_scaled`.
    *   These scaled versions are crucial for algorithms sensitive to feature magnitudes.

7.  **Feature Engineering (Step 12):**
    *   Two new features, 'Runs_Per_Match' and 'Wickets_Per_Match', were created by dividing 'Runs' and 'Wickets' by 'Matches', respectively. This provides a per-match average, potentially offering better insights into player efficiency. Zero division was handled by setting the result to 0.

8.  **Data Type Conversion (Step 13):**
    *   Current data types were reviewed. All numerical features were already in appropriate `int64` or `float64` types. Identifier columns (`PlayerID`, `Name`) remained `object` type, which is suitable. No explicit type conversions were deemed necessary.

9.  **Inconsistent and Noisy Data (Step 14):**
    *   A review confirmed that missing values and duplicates were already handled. Outliers were considered legitimate. Categorical inconsistencies were resolved by encoding. Feature scaling helped manage numerical noise.

10. **Dataset Splitting (Step 15):**
    *   The `df_processed` dataset was split into training and testing sets (`X_train`, `X_test`, `y_train`, `y_test`) using an 80/20 ratio. This prepares the data for model training and evaluation, ensuring that the model is tested on unseen data.

**Overall Impact:** The dataset has been transformed from its raw state into a clean, structured, and numerically represented format, suitable for machine learning tasks. Categorical variables are encoded, numerical features are scaled, new informative features are engineered, and the data is split for robust model development.
